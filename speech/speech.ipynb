```python
import speech_recognition as sr

# Initialize the recognizer
recognizer = sr.Recognizer()
ac_no=input("Enter the Account Number")
# Record audio from microphone
with sr.Microphone() as source:
    print("Say your Code...")
    audio = recognizer.listen(source)

# Convert speech to text
try:
    
    print("Recognizing...")
    text = recognizer.recognize_google(audio)
    print("You said:", text)
    f=open('data/'+ac_no+".txt",'w')
    f.write(text)
    f.close()
except sr.UnknownValueError:
    print("Sorry, could not understand audio.")
except sr.RequestError as e:
    print("Could not request results. Check your internet connection.", e)
```

```python
import speech_recognition as sr
recognizer = sr.Recognizer()
while True:
    f=open('task.txt','r')
    task=f.read()
    f.close()
    if task!='':
        with sr.Microphone() as source:
            print("Say your Code...")
            audio = recognizer.listen(source)
            try:
    
                print("Recognizing...")
                text = recognizer.recognize_google(audio)
                print("You said:", text)
                f=open('data/'+task+".txt",'r')
                code=f.read()
                f.close()
                print(code,text)
                if(text==code):
                    print("matching")
                    f=open('result.txt','w')
                    f.write('matching')
                    f.close()
                else:
                    print('not matching')
                    f=open('result.txt','w')
                    f.write('not matching')
                    f.close()
                    
            except sr.UnknownValueError:
                print("Sorry, could not understand audio.")
            except sr.RequestError as e:
                print("Could not request results. Check your internet connection.", e)
            f=open('task.txt','w')
            f.write('')
            f.close()

            
```
